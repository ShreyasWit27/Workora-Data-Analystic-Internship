```python
import pandas as pd

df = pd.read_csv('customer_churn_raw_v2.csv')

print("CSV file loaded successfully!")
```

```python
print("First 10 Rows:")
display(df.head(10))

print("Dataset Shape:")
print(df.shape)

print("Column Names:")
print(df.columns.tolist())

print("Dataset Information:")
df.info()
```

```python

print("Missing Values in Each Column:")
print(df.isnull().sum())

print("\nTotal Missing Values:")
print(df.isnull().sum().sum())
```

```python
import numpy as np
import pandas as pd

df['TotalChargesINR'] = df['TotalChargesINR'].replace(
    'unknown', np.nan
)

df['TotalChargesINR'] = pd.to_numeric(
    df['TotalChargesINR'], errors='coerce'
)

print("Unknown values converted into NaN successfully!")
```

```python
df['Region'] = df['Region'].fillna('Unknown')

print("Missing Region values handled!")
```

```python
median_value = df['TotalChargesINR'].median()

print("Median Total Charges:", median_value)

df['TotalChargesINR'] = df['TotalChargesINR'].fillna(
    median_value
)

print("Missing TotalChargesINR values handled!")
```

```python
print("Missing Values After Cleaning:")

print(df.isnull().sum())

print("\nTotal Missing Values After Cleaning:")

print(df.isnull().sum().sum())
```

```python
duplicate_count = df.duplicated().sum()

print("Total Duplicate Records:", duplicate_count)
```

```python
duplicates = df[df.duplicated(keep=False)]

print("Duplicate Records:")
display(duplicates)
```

```python
df = df.drop_duplicates()

print("Duplicate records removed successfully!")
```

```python
print("Duplicates After Cleaning:", df.duplicated().sum())

print("Updated Dataset Shape:", df.shape)
```

```python
print("Current Data Types:")
print(df.dtypes)
```

```python

df['Age'] = pd.to_numeric(df['Age'], errors='raise').astype('int64')

df['TenureMonths'] = pd.to_numeric(df['TenureMonths'], errors='raise').astype('int64')

df['MonthlyChargesINR'] = pd.to_numeric(df['MonthlyChargesINR'], errors='raise').astype('float64')

df['TotalChargesINR'] = pd.to_numeric(df['TotalChargesINR'], errors='raise').astype('float64')

df['SupportTickets'] = pd.to_numeric(df['SupportTickets'], errors='raise').astype('int64')

df['LatePayments'] = pd.to_numeric(df['LatePayments'], errors='raise').astype('int64')

categorical_columns = [
    'CustomerID',
    'Region',
    'SubscriptionType',
    'ContractType',
    'PaymentMethod'
]

for col in categorical_columns:
    df[col] = df[col].astype('string')

print("Data types corrected successfully!")
```

```python
print("AutoPay Values:")
print(df['AutoPay'].value_counts(dropna=False))

print("\nChurn Values:")
print(df['Churn'].value_counts(dropna=False))
```

```python
mapping = {
    'Yes': True,
    'No': False
}

df['AutoPay'] = df['AutoPay'].map(mapping).astype('boolean')

df['Churn'] = df['Churn'].map(mapping).astype('boolean')

print("AutoPay and Churn converted successfully!")
```

```python
print("Final Data Types:")
print(df.dtypes)

print("\nDataset Information:")
df.info()
```

```python
import re

df.columns = [
    re.sub(r'[^a-zA-Z0-9]+', '_', col.strip()).strip('_').lower()
    for col in df.columns
]

print("Standardized Column Names:")
print(df.columns.tolist())
```

```python
for col in df.select_dtypes(include=['object', 'string']).columns:
    df[col] = df[col].str.strip()

print("Unnecessary spaces removed!")
```

```python
categorical_columns = [
    'region',
    'subscriptiontype',
    'contracttype',
    'paymentmethod'
]

for col in categorical_columns:
    df[col] = df[col].str.strip().str.title()

print("Categorical values standardized!")
```

```python
print("AutoPay values:")
print(df['autopay'].value_counts(dropna=False))

print("\nChurn values:")
print(df['churn'].value_counts(dropna=False))
```

```python
mapping = {
    'yes': 'Yes',
    'no': 'No'
}

df['autopay'] = df['autopay'].astype('string').str.strip().str.lower().map(mapping)

df['churn'] = df['churn'].astype('string').str.strip().str.lower().map(mapping)
```

```python
print("Final Column Names:")
print(df.columns.tolist())

print("\nUnique Categorical Values:")

for col in ['region', 'subscriptiontype', 'contracttype', 'paymentmethod']:
    print(f"\n{col}:")
    print(df[col].unique())
```

```python
numerical_columns = df.select_dtypes(include=['int64', 'float64']).columns

print("Numerical Columns:")
print(numerical_columns.tolist())
```

```python
outlier_summary = []

for col in numerical_columns:

    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)

    IQR = Q3 - Q1

    lower_limit = Q1 - 1.5 * IQR
    upper_limit = Q3 + 1.5 * IQR

    outliers = df[
        (df[col] < lower_limit) |
        (df[col] > upper_limit)
    ]

    outlier_summary.append({
        'Column': col,
        'Lower Limit': lower_limit,
        'Upper Limit': upper_limit,
        'Outlier Count': len(outliers)
    })

outlier_df = pd.DataFrame(outlier_summary)

display(outlier_df)
```

```python
import matplotlib.pyplot as plt

for col in numerical_columns:

    plt.figure(figsize=(7, 3))

    plt.boxplot(df[col].dropna(), vert=False)

    plt.title(f'Boxplot of {col}')
    plt.xlabel(col)

    plt.show()
```

```python
df_capped = df.copy()

for col in numerical_columns:

    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)

    IQR = Q3 - Q1

    lower_limit = Q1 - 1.5 * IQR
    upper_limit = Q3 + 1.5 * IQR

    df_capped[col] = df[col].clip(
        lower=lower_limit,
        upper=upper_limit
    )

print("IQR-based capping comparison completed!")
```
